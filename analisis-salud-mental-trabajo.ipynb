```python
import re
# Función para convertir rango salarial, la realice en base que ignore espacios, reconozca valores únicos y devuelva solo si realmente no hay números.

def convertir_rango_salarial(rango):
    if pd.isnull(rango):
        return np.nan
    # Eliminar espacios y buscar números
    rango = rango.replace(" ", "")
    numeros = re.findall(r'\d+', rango)

    if len(numeros) == 2:
        minimo = int(numeros[0]) * 1000
        maximo = int(numeros[1]) * 1000
        return (minimo + maximo) / 2
    elif len(numeros) == 1:
        return int(numeros[0]) * 1000
    else:
        return np.nan

df['Salario_medio'] = df['Rango Salarial'].apply(convertir_rango_salarial)
```

# 12- MAPA COROPLETICO:

Realizo la creación de un mapa Coropletico:

Como primera instancia voy a agrupar los datos por País y calcular el promedio de Nivel de estres para cada uno.

Luego voy a utilizar una librería de visualización que sea buena para mapas. Plotly Express es una opción excelente y fácil de usar para mapas coropléticos interactivos.

```python
# Agrupar los datos por País y calcular el promedio de 'Nivel de estres'
df_estres_pais = df.groupby('Pais')['Nivel de Stres'].mean().reset_index()

# Renombrar la columna para claridad en el mapa
df_estres_pais.rename(columns={'Nivel de Stres': 'Promedio de Estres'}, inplace=True)

# Imprimir los primeros resultados para verificar
print("Datos de promedio de estrés por país:")
print(df_estres_pais.head())

# 2. Crear el mapa coroplético usando Plotly Express
fig = px.choropleth(df_estres_pais,
                    locations="Pais",         # Columna que contiene los nombres o códigos de país (corrected to 'Pais')
                    locationmode="country names", # Le decimos a Plotly que 'locations' son nombres de países
                    color="Promedio de Estres", # Columna que se usará para colorear los países
                    hover_name="Pais",        # Columna que aparecerá al pasar el ratón por encima (corrected to 'Pais')
                    color_continuous_scale=px.colors.sequential.Plasma, # Escala de colores
                    title="Promedio de Nivel de Estrés por País", # Título del mapa
                    labels={'Promedio de Estres': 'Nivel de Estrés Promedio'} # Etiqueta de la leyenda
                   )

# Mostrar el mapa
fig.show()
```

**14- "C" Anova:**

Aunque el gráfico anterior sugiere que 'todos los departamentos son iguales' en cuanto al nivel de agotamiento, en ciencia de datos eso no es suficiente. Por eso, decidí calcular un valor P que lo respalde estadísticamente.

Para ello, utilicé la librería Scipy.stats, que es un estándar en análisis estadístico. El código agrupa los datos de 'Nivel de Agotamiento' según el Departamento y realiza una comparación simultánea entre todos los grupos.

```python
grupos = [grupo['Nivel de Agotamiento'].dropna() for nombre, grupo in df.groupby('Departamento')]

#Ejecuto el ANOVA de una vía (One-Way ANOVA)
f_stat, p_value = stats.f_oneway(*grupos)

#Imprimo los resultados de forma legible
print(f"Estadístico F: {f_stat:.4f}")
print(f"Valor P (P-value): {p_value:.4f}")

# Interpretación automática, para ayudarme
alpha = 0.05 # Nivel de significancia estándar (5%)
if p_value < alpha:
    print("CONCLUSIÓN: Hay diferencias significativas. Al menos un departamento es distinto a los demás.")
else:
    print("CONCLUSIÓN: No hay diferencias significativas. El agotamiento es igual en todos los departamentos.")
```

**Explicación del resultado:**

Se llevó a cabo un análisis de varianza (ANOVA) de un factor para evaluar si existen diferencias significativas en el nivel medio de agotamiento entre los distintos departamentos. Los resultados arrojaron un estadístico F de 0.4347 y un p-valor de 0.8246. Dado que el p-valor es considerablemente superior al nivel de significancia establecido (α=0.05α=0.05).

Esto confirma estadísticamente que no existe una asociación significativa entre el departamento laboral y el nivel de agotamiento; es decir, el comportamiento de la variable es homogéneo a través de todas las áreas funcionales.

# 18- PIPELINE DE PREPROCESAMIENTO CON COLUMNTRANSFORMER:

Acá aseguro que el modelo sea robusto, dividido las 8 columnas en dos grupos lógicos (categoricas y númericas).
Entonces:

1- Separo las variables en dos grupos, porque no se puede tratar igual a un número que a una categoría:

**Variables Numéricas:** Como la edad o las horas de trabajo.

**Variables Categóricas**: Como el género o si trabaja remoto.

2. Los Transformadores (La "Limpieza" de Datos): Luego, creo dos Pipelines de procesamiento.

**Para los números (numeric_transformer)**:

- Imputer: Si falta algún dato (un hueco), lo rellenamos con la mediana. Esto evita que el modelo falle por falta de información.

- Scaler: Normalizamos los datos. Por ejemplo, la 'Edad' suele ser un número entre 20-60 y las 'Horas' entre 30-50. El escalado hace que todos los números compitan en igualdad de condiciones.

**Para las categorías (categorical_transformer)**:

- Imputer: Si falta un dato, usamos el valor que más se repite (moda).

- OneHotEncoder: Este paso es vital. Convierte palabras en columnas de 0 y 1. Si el modelo ve 'Hombre/Mujer', lo transforma en algo matemático que el algoritmo pueda procesar.

3. **El ColumnTransformer**:

Su función es saber a qué columna aplicarle cada proceso. Él se encarga de que las columnas numéricas vayan a la de números y las categoricas a la de categorias, y luego vuleve a unir todo en una sola tabla limpia.

4. **Pipeline Final**:

Acá es donde uno todo. Creo un full_model que conecta nuestro Preprocesador directamente con nuestro modelo Random Forest.

¿Por qué hago esto? Por tres razones:

- Orden: El código es más limpio.

- Seguridad: Evitamos errores humanos al procesar los datos de prueba.

- Consistencia: Lo mismo que le hagamos a los datos de entrenamiento, se le aplicará automáticamente a cualquier dato nuevo que llegue en el futuro."

5. **Entrenamiento (.fit)**:

Finalmente, ejecuto full_model.fit(X_train, y_train).

En este momento, los datos entran crudos, pasan por la limpieza, se transforman y el Bosque Aleatorio (Random Forest) empieza a aprender los patrones para predecir el 'Nivel de Agotamiento'. Es el momento donde el modelo realmente adquiere su inteligencia.

**Resumen**, lo que he construido no es solo un modelo, es un flujo automatizado de extremo a extremo. Esto garantiza que el modelo de predicción de agotamiento sea robusto, fácil de mantener y esté listo para recibir datos reales.

```python
# Nombres de funciones corregidos basados ​​en columnas X_train
# X_train.columns: ['Genero', 'Edad', 'Años en la Empresa', 'Hs de Trabajo x Semana', 'Trabajo Remoto', 'Satisfaccion Laboral', 'Nivel de Stres', 'Riesgo de Agotamiento']

numeric_features = ["Hs de Trabajo x Semana", "Edad", "Años en la Empresa", "Nivel de Stres", "Satisfaccion Laboral", "Riesgo de Agotamiento"]
categorical_features = ["Genero", "Trabajo Remoto"]

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False)) # Set sparse_output=False
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
# Unimos el preprocesamiento con el modelo RandomForest que cree antes
full_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', clf) # Aquí clf es tu RandomForestRegressor()
])

# ¡Y ahora entreno todo de un solo golpe!
full_model.fit(X_train, y_train)
```

# 19- MODELOS BASE:

En este paso pongo a "competir" a tres algoritmos diferentes para ver cuál entiende mejor el **agotamiento** de los empleados.

Modelos aplicados:

* **Linear Regressor**:(Regresión Lineal): Es el modelo más sencillo. Busca relaciones directas (si suben las horas, sube el estrés).

* **RandomForest:** El modelo de "bosque". Es excelente para capturar relaciones complejas y no lineales.

* **GradientBoosting:** Un modelo muy potente que aprende de sus propios errores de forma iterativa. Suele ser de los más precisos.



```python
# Antes de empezar, imprimo las columnas y las características que espera el preprocesador.
print(f"Columns in X_train: {X_train.columns.tolist()}\n")
print(f"Numeric features expected by preprocessor: {numeric_features}\n")
print(f"Categorical features expected by preprocessor: {categorical_features}\n")

# Selecciono tres perfiles de algoritmos muy distintos:
#Regresión Lineal
#RandomForest:
#GradientBoosting:

models = {
    "Linear": LinearRegression(),
   "RandomForest": RandomForestRegressor(n_estimators=200, random_state=42),
    "GradientBoosting": GradientBoostingRegressor(random_state=42)
}

results = []
for name, estimator in models.items(): #automatizas el trabajo. Para cada modelo:

    pipe = Pipeline(steps=[("preprocessor", preprocessor), #creo un pipeline
                              ("model", estimator)])
    pipe.fit(X_train, y_train) #entreno el modelo
    y_pred = pipe.predict(X_test) #predice. el modelo intenta adivinar el agotamiento de los datos de prueba (30%) que nunca ha visto.
    results.append({
        "Modelo": name,
        "R2": r2_score(y_test, y_pred), #metrica de precision
        "RMSE": np.sqrt(mean_squared_error(y_test, y_pred)), #raiz de error de cuadratico medio
        "MAE": mean_absolute_error(y_test, y_pred) #error absoluto medio
    })

pd.DataFrame(results).sort_values(by="R2", ascending=False)
```

No hay relaciones "rebuscadas" o patrones ocultos muy complejos que solo un bosque aleatorio podría ver.



**19- "A" R2 AJUSTADO**:

Decidi agregrar el R2 Ajustado ya que a diferencia del R² común, el R² ajustado penaliza la inclusión de variables que no aportan valor real al modelo.

```python
# Calcular el R^2 ajustado
n = X_test.shape[0]  # número de observaciones
p = 1   # número de predictores (excluyendo el intercepto)

r_squared = r2_score(y_test,y_pred)
r_squared_adjusted = 1 - (1 - r_squared) * (n - 1) / (n - p - 1)

print("R^2 ajustado:", r_squared_adjusted)

# tener presente que R2 ajustado es siempre menor que o igual a R2
```

**19- "B" FACTOR DE INFLACIÓN DE LA VARIANZA (VIF)**:

Sirve para detectar la multicolinealidad. Si dos variables independientes están muy correlacionadas entre sí (ej. "RIESGO DE AGOTAMIENTO" y "NIVEL DE ESTRÉS"), el modelo no puede distinguir el efecto individual de cada una.

   

```python
#Ajusto el preprocesador a X_train antes de transformar
preprocessor.fit(X_train)
X_processed = preprocessor.transform(X_train)

#Obtener nombres de funciones después del preprocesamiento
#Si se usa un ColumnTransformer con OneHotEncoder, esto ayuda a obtener nombres legibles
feature_names = preprocessor.get_feature_names_out()
X_processed_df = pd.DataFrame(X_processed, columns=feature_names)

# Calcular VIF para cada característica
vif_data = pd.DataFrame()
vif_data["feature"] = X_processed_df.columns

#Es importante usar los valores .values ​​para variance_inflation_factor
# e iterar por índice. Además, asegúrese de que no haya columnas constantes después del preprocesamiento si se agrega una constante explícita.
# Statsmodels variance_inflation_factor puede manejar una matriz de diseño sin agregar una constante explícitamente, si corresponde.
# Si existe una multicolinealidad perfecta (por ejemplo, al codificar todas las categorías con un único valor sin descartar ninguna),
# VIF podría resultar en infinito, lo cual es un resultado esperado que indica una multicolinealidad severa.

vif_data["VIF"] = [variance_inflation_factor(X_processed_df.values, i)
                   for i in range(len(X_processed_df.columns))]

print("Factor de Inflación de la Varianza (VIF) por característica:")
print(vif_data.sort_values(by="VIF", ascending=False).reset_index(drop=True))
```

**Explicacion del resultado:** Para garantizar la fiabilidad de los coeficientes estimados en el modelo de regresión, se realizó un análisis de Multicolinealidad mediante el Factor de Inflación de la Varianza (VIF). Los resultados obtenidos muestran valores para todas las variables entre 1.00, situándose muy por debajo del umbral crítico de 5. Esto confirma que no existe redundancia entre las variables independientes (como el Nivel de Estrés, Satisfacción Laboral y Riesgo de Agotamiento), permitiendo una interpretación clara del impacto individual de cada factor en la salud mental de los trabajadores.

# 20- OPTIMIZACIÓN CON GRILLAS DE HIPERPARÁMETROS

Tras identificar las limitaciones del modelo lineal, específicamente la presencia de heterocedasticidad confirmada mediante la prueba de White, se procedió a la implementación de modelos de aprendizaje supervisado de mayor complejidad. El objetivo fue maximizar la capacidad predictiva y capturar posibles relaciones no lineales en los factores de salud mental. Pese a que la regresión lineal obtuvo el mejor desempeño inicial, se realizó una optimización exhaustiva de los modelos Random Forest y Gradient Boosting para determinar si, mediante el ajuste fino de hiperparámetros, era posible superar la línea base. Para este proceso se emplearon:

**GridSearchCV**: para una búsqueda exhaustiva en el espacio de parámetros.

**HalvingGridSearchCV**: para optimizar la eficiencia computacional, aplicando en ambos casos una validación cruzada de 5 pliegues (5-folds) para garantizar la robustez de los resultados.

Los cuales estan detallados y realizados a continuación.

**20- "A" GridSearchCV**

```python
# Creo un pipeline. meto el preprocesador dentro de la búsqueda
# and the RandomForestRegressor model
pipeline_rf = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(random_state=42)) # Using a base model with a random_state for reproducibility
])

# defino hiperparametros
param_grid = {
    'model__n_estimators': [100, 200],  # Número de árboles en el bosque
    'model__max_depth': [2, 4, 6],      # Profundidad máxima de cada árbol
    # 'model__min_samples_split': [2,4,6],  # Número mínimo de muestras requeridas para dividir un nodo interno
    # 'model__min_samples_leaf': [2, 4, 6],  # Número mínimo de muestras requeridas en cada hoja del árbol
    # 'model__bootstrap': [True, False]  # Indica si se debe realizar bootstrap para construir los árboles
}

# Utilize the pipeline with GridSearchCV y validacion cruzada
model_RF = GridSearchCV(
    pipeline_rf,
    cv=5,
    param_grid=param_grid,
    scoring='neg_root_mean_squared_error',
    n_jobs=-1,
    verbose=3
)
```

```python
%%time
#Entrenamos nuestro modelo de RF con la grilla ya definida y CV con tamaño de Fold=5
model_RF.fit(X_train, y_train)
```

**Explicación del resultado:** con ese codigo mido exactamente cuánto tiempo tarda la computadora en procesar toda esa celda.

Tengo 6 combinaciones en la grilla (2 de n_estimators × 3 de max_depth).

Configure un CV=5 (Validación Cruzada de 5 pliegues).

6 combinaciones × 5 vueltas = 30 ejecuciones del modelo.

Cuando ejecuto el .fit en nuestra búsqueda de grilla (GridSearchCV), la computadora realiza los siguientes pasos de forma automática:

- El Torneo (Cross-Validation): "Para cada combinación de parámetros, el modelo divide nuestros datos en 5 partes. Entrena con 4 partes y se pone a prueba con la quinta. Repite esto 5 veces rotando las partes. Esto asegura que el resultado no sea por suerte.

- La Evaluación: Calcula el error promedio (RMSE) de cada combinación para ver cuál es la más precisa.

- La Selección del Ganador:Una vez que termina las 30 pruebas, identifica qué combinación de hiperparámetros dio el error más bajo.

- El Re-entrenamiento Final (Refit):Este es el paso más importante: una vez que encuentra los mejores parámetros, el GridSearchCV vuelve a entrenar el modelo por última vez usando la mejor configuración y el 100% de los datos de entrenamiento (X_train). El resultado es un modelo optimizado y listo para usar".   

```python
#Ahora sí ya estamos en condición de realizar nuestras predicciones
prediction = model_RF.predict(X_test)
```

**Explicación de lo Realizado:** Aca ocurren tres cosas automáticas de forma secuencial gracias a que use un Pipeline:

- Transformación Automática: Los datos "crudos" de X_test pasan por el preprocessor. Se rellenan los nulos (imputación), se escalan los números y se codifican las categorías (One-Hot Encoding) exactamente igual a como se hizo con los de entrenamiento.

- Uso del "Mejor" Modelo: Como model_RF viene de un GridSearchCV, no usa cualquier bosque; utiliza automáticamente la combinación de parámetros que resultó ganadora en el paso anterior.

- Generación de Resultados: El modelo analiza las 8 características de los 900 empleados del set de prueba y genera una lista de 900 números, que representan el Nivel de Agotamiento estimado.

```python
# Calculate Root Mean Squared Error (RMSE) for regression
rmse = np.sqrt(mean_squared_error(y_test, prediction))
print(f'RMSE: {rmse:.4f}')
```

**20- "B" Halving Grid Search:**

Al observar en el gráfico anterior que el modelo estaba simplificando demasiado la realidad, decidi hacer una búsqueda mucho más profunda y ambiciosa. Como ahora probamos muchas más combinaciones y parámetros más complejos, use HalvingGridSearchCV.

```python
#creo a pipeline,una canalizacion que incluya el preprocesador y el modelo RandomForestRegressor model

pipeline_rf_halving = Pipeline(steps=[
("preprocessor", preprocessor),
("model", RandomForestRegressor(random_state=42)) # Usando el modelo base con un estado aleatorio para reproducibilidad
])
#Define Hyperparameters for the model within the pipeline
#The keys for the parameters need to be prefixed with the model step name (e.g., 'model__n_estimators')

param_grid = {
'model__n_estimators': [400,500,600],  # Número de árboles en el bosque
'model__max_depth': [18, 20,22],  # Profundidad máxima de cada árbol
'model__min_samples_split': [5, 15],  # Número mínimo de muestras requeridas para dividir un nodo interno
'model__min_samples_leaf': [4, 6]  # Número mínimo de muestras requeridas en cada hoja del árbol
#'model__bootstrap': [True, False]  # Indica si se debe realizar bootstrap para construir los árboles
}
#Aplicamos la grilla al modelo utilizando la pipeline

RF_halving_cv = HalvingGridSearchCV(
pipeline_rf_halving,
param_grid=param_grid,
factor=3,
min_resources=40,
cv=5,
scoring='neg_root_mean_squared_error',
n_jobs=-1,
verbose=2
)
```

**Explicación de lo realizado**: Después de configurar el Halving Grid Search, verificamos el tamaño de nuestro set de entrenamiento con len(X_train). Esto es importante porque el algoritmo de Halving funciona asignando datos de forma progresiva.


```python
%%time
#Entrenamos RF con la grilla definida arriba y CV con tamaño de Fold=5
RF_halving_cv.fit(X_train, y_train)
```

**Explicación de lo realizado:** En el entrenamiento anterior, el modelo estaba limitado (era 'chato'). En este entrenamiento, le estamos permitiendo al modelo volverse experto.

Use %%time para demostrar la eficiencia del método Halving. A pesar de que estamos explorando un espacio de parámetros mucho más complejo que al principio, el tiempo de ejecución es optimizado gracias a que descartamos rápidamente las combinaciones que no funcionan

```python
print("Mejores parametros", RF_halving_cv.best_params_)
print("Mejor CV score", RF_halving_cv.best_score_)

# Calculate predictions on the test set
predictions = RF_halving_cv.predict(X_test)

# Calculate Root Mean Squared Error (RMSE) for regression
rmse = np.sqrt(mean_squared_error(y_test, predictions))
print(f'RMSE del modelo = {round(rmse, 5)}')
```

**Explicación del resultado:** Se determinó que una selva de 600 árboles es ideal. Al ser un número elevado, el modelo reduce la varianza y asegura que las predicciones sean estables, evitando que un solo árbol con ruido afecte el resultado final.

Una profundidad de 22 niveles indica que el fenómeno estudiado tiene una complejidad moderada-alta, requiriendo que los árboles se ramifiquen lo suficiente para capturar patrones detallados en los datos.

Control de Overfitting (min_samples_leaf y min_samples_split): Al establecer un mínimo de 6 muestras por hoja y 15 para dividir un nodo, el modelo evita crear reglas demasiado específicas para pocos datos. Esto actúa como un mecanismo de regularización, forzando al modelo a aprender patrones generales en lugar de memorizar casos particulares (ruido).

Este es el punto más importante para demostrar que el modelo funciona:

-	Error promedio: El modelo presenta un RMSE de 1.5059. Esto significa que, en promedio, las predicciones del modelo fallan por aproximadamente 1.5 unidades.

-	Consistencia CV vs Test: Es fundamental observar que el Mejor CV Score (-1.54) y el RMSE del modelo (1.50) son muy similares. Lo que garantiza que no hay overfitting.
